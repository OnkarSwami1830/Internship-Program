# Dataset Analysis
This notebook inspects the sample job dataset, checks data quality, and summarizes key statistics for the AI job description project.

```python
import pandas as pd

df = pd.read_csv("raw_jobs.csv")
print("Shape:", df.shape)
print("Columns:", list(df.columns))
print(df.head())
```

```python
missing = df.isnull().sum()
print("Missing values:\n")
print(missing)

duplicates = df.duplicated(subset=["job_title", "company", "job_description"]).sum()
print("Duplicate jobs:", duplicates)
```

```python
df["text_length"] = df["job_description"].str.len()
print(df["text_length"].describe())
print("\nAverage text length:", round(df["text_length"].mean(), 2))
```

```python
print("\nRelevant columns:")
print(df[["job_id", "job_title", "company", "location", "job_description", "experience", "education", "salary", "job_type", "text_length"]].head(3).to_string(index=False))
```
