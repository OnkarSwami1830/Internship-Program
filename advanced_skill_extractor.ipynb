# Day 8 - Advanced Skill Extraction

This notebook demonstrates Regex-based normalization and spaCy PhraseMatcher for extracting structured skills from job descriptions.

```python
import re

skill_patterns = {
    "Python": r"\bpython(?:\s*3|3|\s*programming)?\b",
    "SQL": r"\bsql\b",
    "Power BI": r"\bpower\s*bi\b",
    "Tableau": r"\btableau\b",
    "Excel": r"\b(ms\s*excel|excel)\b",
    "AWS": r"\baws\b",
    "Machine Learning": r"\bmachine learning\b",
    "Deep Learning": r"\bdeep learning\b",
    "Natural Language Processing": r"\bnatural language processing\b",
    "Data Science": r"\bdata science\b"
}

def extract_skills(text):
    found = []
    for skill, pattern in skill_patterns.items():
        if re.search(pattern, text.lower()):
            found.append(skill)
    return found

sample_text = "We are looking for a Data Scientist with Python 3, Machine Learning, SQL, Power BI and AWS experience. Knowledge of Natural Language Processing is preferred."
print("Regex extraction:", extract_skills(sample_text))
```

```python
import spacy
from spacy.matcher import PhraseMatcher

try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("spaCy model not found. Install with: python -m spacy download en_core_web_sm")
    nlp = None

if nlp is not None:
    matcher = PhraseMatcher(nlp.vocab, attr="LOWER")
    skills = [
        "Machine Learning",
        "Deep Learning",
        "Natural Language Processing",
        "Power BI",
        "Data Science"
    ]
    patterns = [nlp.make_doc(skill) for skill in skills]
    matcher.add("SKILLS", patterns)

    doc = nlp("Experience in Machine Learning and Power BI is required.")
    matches = matcher(doc)
    result = [doc[start:end].text for _, start, end in matches]
    print("PhraseMatcher result:", result)
```

```python
# Variation mapping examples
tests = [
    "Python",
    "Python3",
    "Python 3",
    "Python Programming",
    "MS Excel",
    "power bi",
    "Data Science"
]

for item in tests:
    print(item, '->', extract_skills(item))
```

## Conclusion

The regex-based extractor normalizes skill variations, while the spaCy PhraseMatcher boosts detection for multi-word technical phrases. Together they improve standardization and job-skill matching quality.
