# Day 6 - NLP Text Preprocessing

This notebook demonstrates the required preprocessing steps for job description text: tokenization, lowercasing, stopword removal, lemmatization, stemming comparison, and sentence segmentation.

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download("punkt", quiet=True)
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)

text = "We are looking for candidates with strong experience in Python and SQL."

tokens = word_tokenize(text)
print("Tokenization:", tokens)
```

```python
lower_tokens = [w.lower() for w in tokens]
print("Lowercase:", lower_tokens)
```

```python
stop_words = set(stopwords.words("english"))
filtered = [w for w in lower_tokens if w.isalnum() and w not in stop_words]
print("Stopword Removal:", filtered)
```

```python
try:
    import spacy
    nlp = spacy.load("en_core_web_sm")
    lemmas = [token.lemma_ for token in nlp(" ".join(filtered))]
except Exception:
    lemmatizer = WordNetLemmatizer()
    lemmas = [lemmatizer.lemmatize(word) for word in filtered]

print("Lemmatization:", lemmas)
```

```python
stemmer = PorterStemmer()
stems = [stemmer.stem(word) for word in filtered]
print("Stemming:", stems)

sentences = sent_tokenize(text)
print("Sentence Segmentation:", sentences)
```

## Summary

This preprocessing pipeline converts raw job descriptions into meaningful tokens that can later be used in skill extraction and classification models.
