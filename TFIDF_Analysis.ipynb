# Day 9 - TF-IDF Analysis

This notebook demonstrates how TF-IDF identifies important keywords in cleaned job descriptions.

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

df = pd.read_csv("raw_jobs.csv")
df["clean_description"] = df["job_description"].fillna("").astype(str)

vectorizer = TfidfVectorizer(max_features=1000, stop_words="english")
X = vectorizer.fit_transform(df["clean_description"])
terms = vectorizer.get_feature_names_out()
scores = X.mean(axis=0).A1

tfidf_df = pd.DataFrame({
    "term": terms,
    "score": scores
})

tfidf_df = tfidf_df.sort_values(by="score", ascending=False)
print(tfidf_df.head(10).to_string(index=False))
```

## Observation
TF-IDF highlights the most statistically important terms such as Python, SQL, AWS, Power, Tableau, and React, helping uncover important job skills automatically.
