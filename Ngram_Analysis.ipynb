# Day 10 - N-Gram Analysis

This notebook demonstrates unigrams, bigrams, and trigrams to detect multi-word technical skills and phrases in job descriptions.

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

df = pd.read_csv("raw_jobs.csv")
df["clean_description"] = df["job_description"].fillna("").astype(str)

vectorizer = TfidfVectorizer(ngram_range=(1,3), stop_words="english", max_features=1000)
X = vectorizer.fit_transform(df["clean_description"])
terms = vectorizer.get_feature_names_out()
scores = X.mean(axis=0).A1

ngram_df = pd.DataFrame({
    "ngram": terms,
    "score": scores
})

ngram_df = ngram_df.sort_values(by="score", ascending=False)
print(ngram_df.head(20).to_string(index=False))
```
