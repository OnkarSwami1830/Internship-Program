# Day 12 - Named Entity Recognition (NER)

This notebook demonstrates default spaCy NER and a custom entity ruler for technical skills, databases, cloud platforms, and tools.

```python
import spacy

nlp = spacy.load("en_core_web_sm")
text = """
Microsoft is hiring a Python Developer
in New York with AWS experience.
"""

doc = nlp(text)
print("Standard NER:")
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
import spacy
from spacy.pipeline import EntityRuler

nlp = spacy.load("en_core_web_sm")
ruler = nlp.add_pipe("entity_ruler", before="ner")
patterns = [
    {"label": "SKILL", "pattern": "Python"},
    {"label": "DATABASE", "pattern": "PostgreSQL"},
    {"label": "CLOUD_PLATFORM", "pattern": "AWS"},
    {"label": "TOOL", "pattern": "Power BI"}
]
ruler.add_patterns(patterns)

custom_text = "Python developer with AWS and PostgreSQL experience."
doc = nlp(custom_text)
print("Custom NER:")
for ent in doc.ents:
    print(ent.text, ent.label_)
```

## Observation
Default spaCy NER identifies organizations and locations, but it does not reliably detect technical skills. Custom entity labels allow the model to recognize Python, AWS, PostgreSQL, and Power BI as domain-specific entities.
