# Day 11 - Word Embeddings & Semantic Similarity

This notebook uses sentence embeddings to compare job text with a list of skills and detect semantically related concepts such as ML -> Machine Learning.

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer("all-MiniLM-L6-v2")

skill_dict = [
    "Python",
    "Machine Learning",
    "SQL",
    "AWS",
    "Power BI"
]

job_text = """
Looking for an ML engineer with Python,
cloud experience and SQL knowledge.
"""

skill_embeddings = model.encode(skill_dict, convert_to_tensor=True)
job_embedding = model.encode(job_text, convert_to_tensor=True)
scores = util.cos_sim(job_embedding, skill_embeddings)

for skill, score in zip(skill_dict, scores[0]):
    print(skill, round(float(score), 3))
```

## Observation
Semantic similarity helps detect contextual relationships that exact string matching misses. For example, ML is close to Machine Learning even when the exact text differs.
